```python
heights = [173, 175, 169, 173, 195,168, 170, float('NaN')]
df['Height'] = heights
df
```

```python
df['Height'] = df['Height'] * 0.01
df
```

```python
weights = df['Weight']
heights = df['Height']
bmi_lst = []
for item in zip(weights, heights):
    mass, height = item
    bmi = mass / (height * height)
    bmi_lst.append(bmi)
print(bmi_lst)


df['BMI'] = bmi_lst
```

```python
def classify_bmi(bmi):
    if bmi >= 30:
        return 'Obesity Range'
    elif bmi < 30 and bmi > 25:
        return 'Overweight Range'
    elif bmi < 25 and bmi >= 18.5:
        return 'Healthy Range'
    elif bmi < 18.5:
        return 'Underweight Range'
```

```python
df
```

```python
df['bmi_class'] = df['bmi'].apply(lambda x: classify_bmi(x))
```
