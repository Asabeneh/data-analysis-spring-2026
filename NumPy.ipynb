```python
import random

def play_game():
    # Set the range and maximum attempts
    low, high = 1, 100
    max_attempts = 7
    secret_number = random.randint(low, high)
    attempts = 0

    print("\n" + "="*40)
    print(f"🎯 WELCOME TO THE NUMBER GUESSING CHALLENGE")
    print(f"I'm thinking of a number between {low} and {high}.")
    print(f"You have {max_attempts} attempts to get it right!")
    print("="*40)

    while attempts < max_attempts:
        try:
            # Get user input and validate it's a number
            guess = int(input(f"\nAttempt {attempts + 1}/{max_attempts} | Enter your guess: "))
        except ValueError:
            print("❌ Invalid input! Please enter a whole number.")
            continue

        attempts += 1

        if guess < secret_number:
            print("📈 TOO LOW! Try a higher number.")
        elif guess > secret_number:
            print("📉 TOO HIGH! Try a lower number.")
        else:
            print(f"\n🎉 CONGRATULATIONS! You found it in {attempts} rounds!")
            return True # Signal a win

    print(f"\n👻 GAME OVER. You've run out of attempts.")
    print(f"The secret number was: {secret_number}")
    return False # Signal a loss

def main():
    while True:
        play_game()
        choice = input("\nWould you like to play again? (y/n): ").lower()
        if choice != 'y':
            print("Thanks for playing! Goodbye.")
            break

if __name__ == "__main__":
    main()
```

```python
def random_id_generator(n = 6):
    letters = list('abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ123456789_')
    rand_id = ''
    for i in range(n):
        letter = np.random.choice(letters)
        rand_id = rand_id + letter
    return rand_id

print(random_id_generator())
print(random_id_generator(16))
print(random_id_generator(24))
print(random_id_generator(5))
```

```python
def random_id_generator2 (n = 6):
    random_id = np.random.choice(list('abcdefghijklmnopkrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'), size= n).tolist()
    return ''.join(random_id)
```
