```python
feedbacks = [
'I love python because it allows me to build any thing.'
'I am not sure about the impact of ai in the society and the future generation.',
'I love ai. It is helping me to be more productive.',
'You must love this course. After completing this course, you will build awesome applications.'
'Life is good...',
'I don\'t like this course at all.',
'I love people',
'I like this project and I would to emulate it.',
'You write feedback and you get sentiment.',
'I hate XYZ',
'I do not like trump',
'I hate Trump!',
'I hate it',
'I don\'t like trump',
'This government is very bad',
'This government is going to be a disaster.',
'I am not expecting anything from this government.',
'Thank you!',
'This training is going to be one of the best and I believe everyone is going learning something valuable.',
'I love people',
'I do not really like it.',
'I don\'t really like'
'I really enjoy todays lesson'
'All the knowledge shall be ours!'
]
```

```python
def determine_sentiment(num):
     if num == 0:
         return 'Neutral'
     elif num < 0:
         return 'Negative'
     elif num > 0:
         return 'Positive'


```

```python
print(len(feedbacks))
```

```python

positives = []
neutrals = []
negatives = []

for feedback in feedbacks:
    blob = TextBlob(feedback)
    polarity = blob.sentiment.polarity
    sentiment = determine_sentiment(polarity)
    if sentiment == 'Positive':
        positives.append(feedback)
    elif sentiment == 'Negative':
        negatives.append(feeback)
    elif sentiment == 'Neutral':
        neutrals.append(feedback)
    # print(determine_sentiment(polarity))


# print('Positives:', len(positives))
# print('Neutral:', len(neutrals))
# print('Negative:', len(negatives))
    
```

```python
fig, ax = plt.subplots()
labels = ['Positive', 'Neutral', 'Negative']
pos  = len(positives)
neu = len(neutrals)
neg = len(negatives)
counts = [pos, neu, neg]
colors = ['tab:green', 'tab:orange', 'tab:red']
ax.bar(labels, counts,  color=colors)
ax.set_ylabel('Counts')
ax.set_xlabel('Sentiment')
ax.set_title('Students Feedback Sentiment Analysis')
plt.show()
```

```python


explode = (0.1, 0, 0)
fig, ax = plt.subplots()
ax.pie(counts, explode=explode, labels=labels,  autopct='%1.1f%%', colors = colors,  shadow=True, startangle=90)
plt.show()
```
